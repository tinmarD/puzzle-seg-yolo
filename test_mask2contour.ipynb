```python
import cv2
import numpy as np
import os
from pathlib import Path
import matplotlib.pyplot as plt 
```

```python
base_dir = r'/home/tinmar/Desktop/Projects/Datasets/Puzzle/tiles'
test_image_name = '12q_1.png'
test_image = os.path.join(base_dir, 'images', test_image_name)
test_mask = os.path.join(base_dir, 'masks_inv', test_image_name)
test_poly = os.path.join(base_dir, 'polygons', test_image_name[:-4] + '.txt')

if not os.path.exists(test_image):
    print(f"Test image does not exist: {test_image}")
if not os.path.exists(test_mask):
    print(f"Test mask does not exist: {test_mask}") 
if not os.path.exists(test_poly):
    print(f"Test polygon file does not exist: {test_poly}")
```

```python
mask = cv2.imread(test_mask, cv2.IMREAD_GRAYSCALE)
image = cv2.imread(test_image)

if mask is None: 
    raise ValueError(f"Failed to load mask image: {test_mask}")

# Ensure binary mask
_, binary = cv2.threshold(mask, 127, 255, cv2.THRESH_BINARY)
h, w = binary.shape

# Find contours
contours, _ = cv2.findContours(binary, cv2.RETR_LIST, cv2.CHAIN_APPROX_SIMPLE)

cv2.drawContours(image, contours, -1, (0, 255, 0), 3)
cv2.imshow('Contours', image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
